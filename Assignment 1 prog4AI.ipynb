```python
#task 2
sentence=input("Enter a sentence:")
words=sentence.split()
total_words=len(words)
print("Total words",total_words)
total_characters=len(sentence)
print("Total characters:",total_characters)
longest_word=max(words,key=len)
print("Largest word",longest_word)
shortest_word=min(words,key=len)
print("shortest word",shortest_word)
vowels="aeiouAEIOU"
vowel_count=0
for character in sentence:
    if character in vowels:
        vowel_count+=1
unique_words=set(words)
alphabetical_word=sorted(unique_words)
print("\n--- WORD ANALYSIS RESULTS ---")
print("1. Total Words:", total_words)
print("2. Total Characters (without spaces):", total_characters)
print("3. Longest Word:", longest_word)
print("4. Shortest Word:", shortest_word)
print("5. Vowel Count:", vowel_count)
print("6. Unique Words Set:", unique_words)
print("7. Alphabetical Order:")
for word in alphabetical_word:
    print(word)
with open("word_analysis.txt", "w") as file:
    file.write(f"Sentence: {sentence}\n")
    file.write(f"Total Words: {total_words}\n")
    file.write(f"Total Characters: {total_characters}\n")
    file.write(f"Longest Word: {longest_word}\n")
    file.write(f"Shortest Word: {shortest_word}\n")
    file.write(f"Vowel Count: {vowel_count}\n")
    file.write(f"Unique Words: {unique_words}\n")
    file.write(f"Alphabetical Order: {alphabetical_word}\n")

print("\nSaved successfully to 'word_analysis.txt'!")
```

```python
#task 3
prices={}
quantities={}
totals={}
def calculate_bill(totals):
    bill=0
    for name in totals:
        bill=bill+totals[name]
    return bill
for i in range(5):
    print("product",i+1)
    name=input("Enter product name:")
    price=int(input("Enter price:"))
    quantity=int(input("Enter quantity:"))
    prices[name]=price
    quantities[name]=quantity
    totals[name]=price*quantity
    bill=calculate_bill(totals)
    if bill>10000:
        discount=bill*10/100
    else:
        discount=0
final_bill=discount
max_price=0
max_name=""
for name in prices:
        if prices[name]>max_price:
            max_price=prices[name]
            max_name=name
print("---Bill---")
for name in totals:
    print(name,prices[name],"x",quantities[name],"=",totals[name])
print("Total:",bill)
print("Discount:",discount)
print("Final Bill:",final_bill)
print("Most expensive product:",max_name)
file=open("bill.text","w")
file.write("---BILL---\n")
file.write(name+""+str(prices[name])+"x"+str(quantities[name])+"="+str(totals[name])+"\n")
file.write("Total:"+str(bill)+"\n")
file.write("Discount:"+str(discount)+"\n")
file.write("Final Bill:"+str(final_bill)+"\n")
file.write("Most Expensive product:"+max_name+"\n")
file.close()
print("Bill saved in bill.txt")
            
```

```python
#task 4
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Error: cannot divide by zero"
    return a / b


def modulus(a, b):
    if b == 0:
        return "Error: cannot divide by zero"
    return a % b


while True:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    print("Choose operation:")
    print("+  Addition")
    print("-  Subtraction")
    print("*  Multiplication")
    print("/  Division")
    print("%  Modulus")
    op = input("Enter operation: ")

    if op == "+":
        result = add(num1, num2)
    elif op == "-":
        result = subtract(num1, num2)
    elif op == "*":
        result = multiply(num1, num2)
    elif op == "/":
        result = divide(num1, num2)
    elif op == "%":
        result = modulus(num1, num2)
    else:
        result = "Invalid operation"

    print("Result:", result)

    again = input("Do you want to calculate again? (yes/no): ")
    if again.lower() != "yes":
        print("Bye!")
        break
```

```python
#task 5
menu={
    "1":"Burger",
    "2":"Pizza",
    "3":"Fries",
    "4":"Drink"
}
prices={
    "Burger":1000,
    "Pizza":2000,
    "Fries":300,
    "Drink":150,
}
def cal_price(item,quantity):
    return prices[item]*quantity
print("---MENU---")
print("1.Burger Rs.1000")
print("2.Pizza Rs. 2000")
print("3.Fries Rs.300")
print("4.Drink Rs.150")
bill=0
order=""
while True:
    choice=input("Select item number")
    if choice in menu:
        item=menu[choice]
        quantity=int(input("Enter quantity:"))
        price=cal_price(item,quantity)
        bill=bill+price
        order=order+item+"x"+str(quantity)+"=Rs."+str(price)+"\n"
    else:
        print("Wrong choice")
        more=input("Want to order more?(yes/no)")
        if more!="yes":
            break
        print("---Your Bill---")
        print(order)
        print("Total Bill= Rs.",bill)
        file=open("orders.txt","w")
        file.write("order")
        file.write("Total Bill=Rs."+str(bill)+"\n")
        file.close()
        print("Order saved in orders.txt")
```
